```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create dataset
data = {
    'tenure': [1, 5, 10, 20, 2, 8, 15, 3, 12, 25],
    'MonthlyCharges': [70, 80, 60, 90, 85, 75, 65, 95, 55, 50],
    'Contract': ['Month-to-month', 'One year', 'Two year', 'Month-to-month', 'Month-to-month',
                 'One year', 'Two year', 'Month-to-month', 'Two year', 'One year'],
    'Churn': ['Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'No', 'Yes', 'No', 'No']
}

df = pd.DataFrame(data)

print("Dataset:\n", df)

# Churn count
print("\nChurn Count:\n", df['Churn'].value_counts())

# Plot
sns.countplot(x='Churn', data=df)
plt.title("Churn Distribution")
plt.show()

sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.show()

sns.countplot(x='Contract', hue='Churn', data=df)
plt.xticks(rotation=30)
plt.show()
```
